```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create vectors
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

print("VECTOR OPERATIONS")
print("Vector 1:", v1)
print("Vector 2:", v2)

# 2. Vector operations
print("\nVector Addition:", v1 + v2)
print("Vector Subtraction:", v1 - v2)
print("Element-wise Multiplication:", v1 * v2)
print("Dot Product:", np.dot(v1, v2))

# Magnitude and normalization
print("Magnitude:", np.linalg.norm(v1))
normalized = v1 / np.linalg.norm(v1)
print("Normalized Vector:", normalized)

# 3. Create matrices
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[6, 5, 4],
              [3, 2, 1]])

print("\nMATRIX A:\n", A)
print("\nMATRIX B:\n", B)

# 4. Matrix operations
print("\nMatrix Addition:\n", A + B)
print("\nMatrix Subtraction:\n", A - B)
print("\nElement-wise Multiplication:\n", A * B)
print("\nTranspose of A:\n", A.T)

# Matrix multiplication
W = np.array([[1, 2],
              [3, 4],
              [5, 6]])

print("\nMatrix Multiplication:\n", A @ W)

# 5. Determinant and inverse
M = np.array([[2, 1],
              [1, 3]])

print("\nDeterminant:", np.linalg.det(M))
print("\nInverse:\n", np.linalg.inv(M))

# 6. Neural network forward operation
input_data = np.array([1, 2, 3])

weights = np.array([[0.2, 0.4],
                    [0.5, 0.1],
                    [0.3, 0.6]])

bias = np.array([0.1, 0.2])

z = np.dot(input_data, weights) + bias

# ReLU activation
output = np.maximum(0, z)

print("\nFORWARD PROPAGATION")
print("Input:", input_data)
print("Weighted Sum:", z)
print("ReLU Output:", output)

# 7. Visualization
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(A, cmap="viridis")
plt.title("Matrix A")
plt.colorbar()

plt.subplot(1, 2, 2)
plt.imshow(W, cmap="plasma")
plt.title("Weight Matrix")
plt.colorbar()

plt.tight_layout()
plt.show()
```
